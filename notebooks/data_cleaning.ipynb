```python
# Import pandas library for data analysis and manipulation
import pandas as pd
```

```python
# Read/load the Excel dataset into a pandas DataFrame
# df is the variable storing the dataset
df = pd.read_excel("retail_sales.xlsx")
# Display first 5 rows of the dataset
df.head()
```

```python
# Display dataset information
df.info()
```

```python
# Convert order_date column from object/text to datetime format
# Convert order_time column from object/text to datetime format

df['order_date'] = pd.to_datetime(df['order_date'])
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S')
```

```python
# Check for missing/null values in all columns
# 0 means no missing values
df.isnull().sum()
```

```python
# Check for duplicate rows in the dataset
# 0 means no duplicate records found
df.duplicated().sum()
```

```python
# cleaned_retail_sales.xlsx is the new cleaned dataset file
# index=False prevents pandas from adding an extra index column

df.to_excel("cleaned_retail_sales.xlsx", index=False)
```
